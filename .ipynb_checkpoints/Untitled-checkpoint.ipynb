```python
morda = [10.5, 11.2, 9.8, None, 11.5, None]
```

```python
count = 0
total = 0

for val in morda:
    if val is not None:
        count += 1
        total += val

average = total / count
average
        
```

```python
morda = [10.5, 11.2, 9.8, None, 11.5, None]

count = 0
total = 0

for val in morda:
    if val is not None:
        count += 1
        total += val

average = total/ count

for idx, val in enumerate(morda):
    if val is None:
        morda[idx] = average
```

```python
morda = [10.5, 11.2, 9.8, None, 11.5, None]

count = sum(1 for val in morda if val is not None)
total = sum(val for val in morda if val is not None)
ave = total / count

data = [val if val is not None else ave for val in morda]
```
